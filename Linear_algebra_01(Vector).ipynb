# Scalar
- A scalar is a number. Examples of scalars are temperature, distance, speed, or mass - all
quantities that have a magnitude but no “direction”, other than perhaps positive or negative.

# Vector

- A vector is just a list of numbers. But those numbers mean something -- they're coordinates in space.
- Let A and B be two points. A directed line segment from A to B. This directed line segment constitutes a vector. If you can move the line segment to another line segment with the same direction and length, they constitute the same vector.
- In AI, vectors represent everything:
    - A word → a vector of 768 numbers (its "meaning" in embedding space)
    - An image → a vector of millions of pixel values
    - A user → a vector of preferences
- A shorter notation for vectors is to use a single bold face characters, such as **v**
- We also use the terms tail point and tip point of a vector when this is convenient, where the tip point is where the arrowhead is, and the tail point is the other end.
- A vector is completely defined by its direction, and its length.
- Length of vector :- ||v||


---

- The length of a vector is a scalar, which just means that it is a regular number, such as 7.5. The term scalar is used to emphasize that it is just a number and not a vector or a point.

- Note that the order of the points is important, i.e., if you change the order of A  and B, another vector, BA , is obtained. It has opposite direction, but the same length.
- Even AA is a vector, which is called the zero vector.
- The zero vector is denoted by AA, and can be created using a directed line segment using the same point twice. The length of the zero vector is zero(||0||=0).


---

- Two vectors,**u** and **v**, are parallel if they have the same direction or opposite directions, but not necessarily the same lengths(u||v).
- The zero vector is said to be parallel to all other vectors.

### Vector from scratch

````python
class Vector:
    def __init__(self, components):
        self.components = list(components)
        self.dim = len(self.components)

    def __add__(self, other):
        return Vector([a + b for a, b in zip(self.components, other.components)])

    def __sub__(self, other):
        return Vector([a - b for a, b in zip(self.components, other.components)])

    def dot(self, other):
        return sum(a * b for a, b in zip(self.components, other.components))

    def magnitude(self):
        return sum(x**2 for x in self.components) ** 0.5

    def normalize(self):
        mag = self.magnitude()
        return Vector([x / mag for x in self.components])

    def cosine_similarity(self, other):
        return self.dot(other) / (self.magnitude() * other.magnitude())

    def __repr__(self):
        return f"Vector({self.components})"


a = Vector([1, 2, 3])
b = Vector([4, 5, 6])

print(f"a + b = {a + b}")
print(f"a · b = {a.dot(b)}")
print(f"|a| = {a.magnitude():.4f}")
print(f"cosine similarity = {a.cosine_similarity(b):.4f}")
````

## Vector Addition (Head-to-Tail Method)

The sum $\vec{u} + \vec{v}$ of two vectors, $\vec{u}$ and $\vec{v}$, is constructed using the following geometric steps:

### Core Properties

* **Arbitrary Placement:** Vector $\vec{u}$ is placed at an arbitrary starting location (typically the origin).
* **Tail-to-Tip Alignment:** Vector $\vec{v}$ is placed such that its **tail point** coincides exactly with $\vec{u}$'s **tip point**.
* **Resultant Vector:** The sum vector $\vec{u} + \vec{v}$ is the direct path that **starts at $\vec{u}$'s tail point** and **ends at $\vec{v}$'s tip point**.


````python
import matplotlib.pyplot as plt
import numpy as np

# Define vector components [x, y]
A = np.array([4, 3])
B = np.array([3, -2])
R = A + B  # Resultant vector [7, 1]

# Set up the plot
fig, ax = plt.subplots(figsize=(8, 8))

# Plot Vector A (starting at origin)
ax.quiver(0, 0, A[0], A[1], angles='xy', scale_units='xy', scale=1,
          color='#1f77b4', label=r'Vector $\vec{A}$')

# Plot Vector B (starting at the tip of A)
ax.quiver(A[0], A[1], B[0], B[1], angles='xy', scale_units='xy', scale=1,
          color='#2ca02c', label=r'Vector $\vec{B}$')

# Plot Resultant Vector R (starting at origin, ending at tip of B)
ax.quiver(0, 0, R[0], R[1], angles='xy', scale_units='xy', scale=1,
          color='#d62728', linestyle='--', facecolor='none', edgecolor='#d62728',
          linewidth=2, label=r'Resultant $\vec{R} = \vec{A} + \vec{B}$')

# Add text labels for the critical geometric points
ax.text(0, -0.4, 'Tail of A\n(Start of R)', ha='center', va='top', fontsize=10, weight='bold')
ax.text(A[0] + 0.2, A[1] + 0.2, 'Tip of A\nTail of B', ha='left', va='bottom', fontsize=10, weight='bold')
ax.text(R[0] + 0.2, R[1], 'Tip of B\n(End of R)', ha='left', va='center', fontsize=10, weight='bold')

# Configure grid, limits, and labels
ax.set_xlim(-1, 9)
ax.set_ylim(-1, 5)
ax.axhline(0, color='black', linewidth=0.8)
ax.axvline(0, color='black', linewidth=0.8)
ax.grid(True, which='both', linestyle=':', alpha=0.5)
ax.set_aspect('equal')

# Title and Legend
ax.set_title('Vector Addition: Head-to-Tail Method', fontsize=14, pad=15, weight='bold')
ax.set_xlabel('X Component', fontsize=12)
ax.set_ylabel('Y Component', fontsize=12)
ax.legend(loc='upper left', fontsize=11)

# Display the plot in Colab
plt.show()
````

### Properties of Vector Arithmetic

Assuming that $\mathbf{u}$, $\mathbf{v}$, and $\mathbf{w}$ are vectors of the same size, and that $k$ and $l$ are scalars, then the following rules hold:

*   **(i)** $\mathbf{u} + \mathbf{v} = \mathbf{v} + \mathbf{u}$ (commutativity)
*   **(ii)** $(\mathbf{u} + \mathbf{v}) + \mathbf{w} = \mathbf{u} + (\mathbf{v} + \mathbf{w})$ (associativity)
*   **(iii)** $\mathbf{v} + \mathbf{0} = \mathbf{v}$ (zero existence)
*   **(iv)** $\mathbf{v} + (-\mathbf{v}) = \mathbf{0}$ (negative vector existence)
*   **(v)** $k(l\mathbf{v}) = (kl)\mathbf{v}$ (associativity)
*   **(vi)** $1\mathbf{v} = \mathbf{v}$ (multiplicative one)
*   **(vii)** $0\mathbf{v} = \mathbf{0}$ (multiplicative zero)
*   **(viii)** $k\mathbf{0} = \mathbf{0}$ (multiplicative zero vector)
*   **(ix)** $k(\mathbf{u} + \mathbf{v}) = k\mathbf{u} + k\mathbf{v}$ (distributivity 1)
*   **(x)** $(k + l)\mathbf{v} = k\mathbf{v} + l\mathbf{v}$ (distributivity 2)

## Adding Three Vectors

### 1. Algebraic Addition
To add three vectors algebraically, we simply add their corresponding components. Let the three vectors be:
* $\mathbf{u} = [u_1, u_2, \dots, u_n]$
* $\mathbf{v} = [v_1, v_2, \dots, v_n]$
* $\mathbf{w} = [w_1, w_2, \dots, w_n]$

The sum $\mathbf{s} = \mathbf{u} + \mathbf{v} + \mathbf{w}$ is given by:
$\mathbf{s} = [u_1 + v_1 + w_1, u_2 + v_2 + w_2, \dots, u_n + v_n + w_n]$

### 2. Geometric Addition (Polygon Law)
To add three vectors geometrically, we use the **head-to-tail method** (also known as the polygon law of vector addition):
1. Place the tail of the first vector $\mathbf{u}$ at the origin.
2. Place the tail of the second vector $\mathbf{v}$ at the head of $\mathbf{u}$.
3. Place the tail of the third vector $\mathbf{w}$ at the head of $\mathbf{v}$.
4. The resultant vector $\mathbf{u} + \mathbf{v} + \mathbf{w}$ is drawn from the tail of $\mathbf{u}$ (the starting point) to the head of $\mathbf{w}$ (the final point).

By the associative property of vector addition, the order in which we add the vectors

````python
# Helper function to add two vectors (lists) element by element
def add_vectors(vec1, vec2):
    return [x + y for x, y in zip(vec1, vec2)]

# Define three vectors as standard Python lists
u = [1, 2, 3]
v = [4, -1, 0]
w = [-2, 5, 2]

# Calculate (u + v) + w
u_plus_v = add_vectors(u, v)
sum1 = add_vectors(u_plus_v, w)

# Calculate u + (v + w)
v_plus_w = add_vectors(v, w)
sum2 = add_vectors(u, v_plus_w)

print(f"(u + v) + w = {sum1}")
print(f"u + (v + w) = {sum2}")

# Check if they are exactly equal (verifying associativity)
print("\nAre they equal?", sum1 == sum2)
````

## Scalar Multiplication of a Vector

When a vector $\vec{a}$ is multiplied by a scalar $c$, the vector $\vec{b} = c\vec{a}$ is obtained.

### Core Properties

* **Parallelism:** The resulting vector $\vec{b}$ is always **parallel** to $\vec{a}$.
* **Magnitude:** The length of the new vector is scaled by the absolute value of the scalar:
  $$\|\vec{b}\| = |c| \cdot \|\vec{a}\|$$
* **Direction:**
  * If **$c > 0$**, $\vec{b}$ points in the **same direction** as $\vec{a}$.
  * If **$c < 0$**, $\vec{b}$ points in the **opposite direction** of $\vec{a}$.
  * If **$c = 0$**, then $\vec{b} = \vec{0}$ (the zero vector).


````python
import matplotlib.pyplot as plt
import numpy as np

# Define base vector a = [2, 1]
a = np.array([2, 1])

# Define scalars
c_pos = 2.5    # Positive scalar (stretches, same direction)
c_neg = -1.5   # Negative scalar (stretches, opposite direction)

# Calculate scaled vectors
b_pos = c_pos * a
b_neg = c_neg * a

# Set up the plot
fig, ax = plt.subplots(figsize=(10, 8))

# 1. Plot Original Vector a
ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1,
          color='#1f77b4', width=0.008, label=r'Original Vector $\vec{a} = [2, 1]$')

# 2. Plot Positively Scaled Vector (c * a) - offset slightly for visual clarity
ax.quiver(0, 0, b_pos[0], b_pos[1], angles='xy', scale_units='xy', scale=1,
          color='#2ca02c', width=0.005, alpha=0.7,
          label=r'Scaled Vector $\vec{b} = 2.5\vec{a}$ (Same direction)')

# 3. Plot Negatively Scaled Vector (c * a)
ax.quiver(0, 0, b_neg[0], b_neg[1], angles='xy', scale_units='xy', scale=1,
          color='#d62728', width=0.005,
          label=r'Scaled Vector $\vec{b} = -1.5\vec{a}$ (Opposite direction)')

# Configure layout, axes, and grid
ax.set_xlim(-4, 6)
ax.set_ylim(-3, 4)
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.grid(True, which='both', linestyle=':', alpha=0.5)
ax.set_aspect('equal')

# Title, Labels, and Legend
# Title, Labels, and Legend
ax.set_title(r'Scalar Multiplication of a Vector: $\vec{b} = c\vec{a}$', fontsize=14, pad=15, weight='bold')
ax.set_xlabel('X Axis', fontsize=12)
ax.set_ylabel('Y Axis', fontsize=12)
ax.legend(loc='upper left', fontsize=11)
# Display the plot
plt.show()
````

## Linear Independence
- Vectors are linearly independent if no vector in the set can be written as a combination of the others. If v1,v2,v3 are independent, they span a 3D space. If one is a combination of the others, they only span a plane.

- Why it matters for AI: your feature matrix should have linearly independent columns. If two features are perfectly correlated (linearly dependent), the model cannot distinguish their effects. This causes multicollinearity in regression -- the weight matrix becomes unstable, and small input changes produce wild output swings.


````python
v1 = [1, 0, 0]
v2 = [0, 1, 0]
v3 = [2, 1, 0]   # v3 = 2*v1 + v2

# In a dataset: if feature_3 = 2*feature_1 + feature_2, adding feature_3 gives the model zero new information.
# Worse, it makes the normal equations singular -- there is no unique solution for the weights.
````

````python
def is_linearly_independent(vectors):
    n = len(vectors)
    dim = len(vectors[0].components)
    mat = Matrix([v.components[:] for v in vectors])
    rows = [row[:] for row in mat.rows]
    rank = 0
    for col in range(dim):
        pivot = None
        for row in range(rank, len(rows)):
            if abs(rows[row][col]) > 1e-10:
                pivot = row
                break
        if pivot is None:
            continue
        rows[rank], rows[pivot] = rows[pivot], rows[rank]
        scale = rows[rank][col]
        rows[rank] = [x / scale for x in rows[rank]]
        for row in range(len(rows)):
            if row != rank and abs(rows[row][col]) > 1e-10:
                factor = rows[row][col]
                rows[row] = [rows[row][j] - factor * rows[rank][j] for j in range(dim)]
        rank += 1
    return rank == n


def project(a, b):
    scalar = a.dot(b) / b.dot(b)
    return Vector([scalar * x for x in b.components])


def gram_schmidt(vectors):
    orthonormal = []
    for v in vectors:
        w = v
        for u in orthonormal:
            proj = project(w, u)
            w = w - proj
        if w.magnitude() < 1e-10:
            continue
        orthonormal.append(w.normalize())
    return orthonormal


v1 = Vector([1, 0, 0])
v2 = Vector([1, 1, 0])
v3 = Vector([1, 1, 1])
basis = gram_schmidt([v1, v2, v3])
for i, u in enumerate(basis):
    print(f"u{i+1} = {u}")
    print(f"  |u{i+1}| = {u.magnitude():.6f}")

print(f"u1 · u2 = {basis[0].dot(basis[1]):.6f}")
print(f"u1 · u3 = {basis[0].dot(basis[2]):.6f}")
print(f"u2 · u3 = {basis[1].dot(basis[2]):.6f}")
````

### Basis and Dimension
A **basis** is a minimal set of linearly independent vectors that span the entire space. The number of basis vectors is the **dimension** of the space.
* The standard basis for 3D space is $\{[1,0,0], [0,1,0], [0,0,1]\}$.
* Any three independent vectors in 3D form a valid basis.
* The choice of basis is a choice of coordinate system.

---

### Matrix Rank
**Rank of a matrix** = number of linearly independent columns = number of linearly independent rows.

If $\text{rank} < \min(\text{rows}, \text{cols})$, the matrix is **rank-deficient**. This means:
* The system has infinitely many solutions (or none)
* Information is lost in the transformation
* The matrix cannot be inverted

---

### Rank Implications in Machine Learning

| Situation | Rank | What it means for ML |
| :--- | :--- | :--- |
| **Full rank** | $\text{rank} = \min(m,n)$ | **Maximum possible:** With at least as many rows as columns (full column rank), the least-squares solution is unique. With fewer rows than columns, infinitely many solutions fit exactly. |
| **Rank deficient** | $\text{rank} < \min(m,n)$ | **Below maximum:** Features are redundant. Infinitely many weight solutions. Regularization needed. |
| **Rank 1** | $1$ | Every column is a scaled copy of one vector. All data lies on a line. |
| **Near rank-deficient** *(small singular values)* | Numerically low | Matrix is ill-conditioned. Tiny input noise causes large output changes. Use SVD truncation or ridge regression. |

### Vector Projection
Projecting vector $\mathbf{a}$ onto vector $\mathbf{b}$ gives the component of $\mathbf{a}$ in the direction of $\mathbf{b}$:

$$ \text{proj}_{\mathbf{b}}(\mathbf{a}) = \frac{\mathbf{a} \cdot \mathbf{b}}{\mathbf{b} \cdot \mathbf{b}} \mathbf{b} $$

The residual $(\mathbf{a} - \text{proj}_{\mathbf{b}}(\mathbf{a}))$ is perpendicular to $\mathbf{b}$. This orthogonal decomposition is the foundation of least-squares fitting.

### Projection is everywhere in ML:
* **Linear regression:** Minimizes the distance from observations to the column space — the solution IS a projection.
* **PCA (Principal Component Analysis):** Projects data onto the directions of maximum variance.
* **Attention in Transformers:** Computes projections of queries onto keys.


### Example
Given $\mathbf{a} = [3, 4]$ and $\mathbf{b} = [1, 0]$:

$$ \text{proj}_{\mathbf{b}}(\mathbf{a}) = \frac{3 \cdot 1 + 4 \cdot 0}{1^2 + 0^2} [1, 0] = \frac{3}{1} [1, 0] = [3, 0] $$

The projection drops the y-component. This is **dimensionality reduction** in its simplest form — throw away the directions you don't care about.

### Gram-Schmidt Process
Converting any set of independent vectors into an **orthonormal basis**. Orthonormal means every vector has length 1 and every pair is perpendicular.

**The Algorithm:**
1. Take the first vector, normalize it.
2. Take the second vector, subtract its projection onto the first, then normalize.
3. Take the third vector, subtract its projections onto all previous vectors, then normalize.
4. Repeat for the remaining vectors.

```text
Input:  v1, v2, v3, ... (linearly independent)

u1 = v1 / |v1|

w2 = v2 - (v2 ⋅ u1) * u1
u2 = w2 / |w2|

w3 = v3 - (v3 ⋅ u1) * u1 - (v3 ⋅ u2) * u2
u3 = w3 / |w3|

Output: u1, u2, u3, ... (orthonormal basis)

### Column Vector and Row Vector

In linear algebra, a vector is typically represented as a matrix with either a single column or a single row.

#### 1. Column Vector
A **column vector** is a matrix with $m$ rows and exactly $1$ column (an $m \times 1$ matrix). In most mathematics and machine learning contexts, when a vector is mentioned without specifying its layout, it is assumed to be a column vector.

**Example (a $3 \times 1$ column vector):**
$$ \mathbf{v} = \begin{bmatrix} 2 \\ 5 \\ -1 \end{bmatrix} $$

#### 2. Row Vector
A **row vector** is a matrix with exactly $1$ row and $n$ columns (a $1 \times n$ matrix).

**Example (a $1 \times 3$ row vector):**
$$ \mathbf{u} = \begin{bmatrix} 2 & 5 & -1 \end{bmatrix} $$

---

### The Transpose Relationship
You can convert a column vector to a row vector (and vice versa) using the **transpose** operation, denoted by a superscript $T$. The transpose flips the matrix over its diagonal.

**Example:**
$$ \mathbf{v}^T = \begin{bmatrix} 2 \\ 5 \\ -1 \end{bmatrix}^T = \begin{bmatrix} 2 & 5 & -1 \end{bmatrix} $$

````python
# 1. Define a column vector as a list of lists (a 3x1 structure)
v_col = [[2],
         [5],
         [-1]]

print("--- Column Vector ---")
for row in v_col:
    print(row)

# 2. Transpose the column vector to a row vector
# We create a new list containing one inner list with all the first elements
v_row = [[item[0] for item in v_col]]

print("\n--- Transposed to Row Vector ---")
print(v_row)

# 3. Define a row vector directly (a 1x3 structure)
u_row = [[10, 20, 30]]

print("\n--- Original Row Vector ---")
print(u_row)

# 4. Transpose the row vector back into a column vector
# We take each element from the first (and only) row and wrap it in its own list
u_col = [[item] for item in u_row[0]]

print("\n--- Transposed to Column Vector ---")
for row in u_col:
    print(row)

# ---------------------------------------------------------
# The Pythonic "One-Liner" for any 2D list transpose
# ---------------------------------------------------------
# Using zip(*...) unpacks the list and groups elements by column
generic_transpose = list(map(list, zip(*v_col)))

print("\n--- Generic Transpose (using zip) ---")
print(generic_transpose)
````
